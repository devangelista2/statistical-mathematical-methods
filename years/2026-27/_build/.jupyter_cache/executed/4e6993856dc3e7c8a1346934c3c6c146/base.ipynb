```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameterized function f
def f(theta, x):
    return theta[0] + theta[1] * x

# Choose two different values for the parameters
theta = (1, 0.2)
theta2 = (0, -1)

# Visualize the model prediction in the range [-5, 5]
xx = np.linspace(-5, 5, 10)
yy = f(theta, xx)
yy2 = f(theta2, xx)

plt.plot(xx, yy, 'o--', label=r"$\Theta = (1, 0.2)$")
plt.plot(xx, yy2, 'o--', label=r"$\Theta = (0, -1)$")
plt.grid()
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"A plot of $f_\Theta(x)$")
plt.show()
```

```python
# Define synthetic datapoints
np.random.seed(42)
x_data = np.linspace(-5, 5, 20)
y_data = 1 + 0.2 * x_data + np.random.normal(0, 0.5, x_data.shape)

# Plot
plt.plot(xx, yy, 'o--', label=r"$\Theta = (1, 0.2)$")
plt.plot(xx, yy2, 'o--', label=r"$\Theta = (0, -1)$")
plt.plot(x_data, y_data, 'o', label="Data")
plt.grid()
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"A plot of $f_\Theta(x)$ against the data")
plt.show()
```

```python
# Define synthetic datapoints
x_data = np.linspace(-5, 5, 20)
y_data = np.sin(x_data)

# Create a linear model approximation
xx = np.linspace(-5, 5, 10)
theta = (0, 0.1)
yy = f(theta, xx)

# Plot
plt.plot(xx, yy, 'o--', label="Linear model")
plt.plot(x_data, y_data, 'o', label="Data")
plt.grid()
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title(r"A linear $f_\Theta(x)$ approximating $\sin(x)$")
plt.show()
```

```python
np.random.seed(0)

# Regression: y is a real number
x_reg = np.random.uniform(-3, 3, 100)
y_reg = np.sin(x_reg) + 0.2 * np.random.normal(0, 1, x_reg.shape)

# Classification: y is either 0 or 1
N = 200
x_cls = np.random.normal(0, 1, (N, 2))
y_cls = (x_cls[:, 0]**2 + x_cls[:, 1]**2 < 1).astype(int)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(x_reg, y_reg, 'o')
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[0].set_title("Regression: $y \\in \\mathbb{R}$")
axes[0].grid()

axes[1].scatter(x_cls[y_cls == 0, 0], x_cls[y_cls == 0, 1], label="Class 0")
axes[1].scatter(x_cls[y_cls == 1, 0], x_cls[y_cls == 1, 1], label="Class 1")
axes[1].set_xlabel("$x_1$")
axes[1].set_ylabel("$x_2$")
axes[1].set_title("Classification: $y \\in \\{0, 1\\}$")
axes[1].set_aspect("equal")
axes[1].legend()
axes[1].grid()
plt.show()
```
