## Project: training a Neural Network on a real dataset

Choose a **classification** dataset, and train on it an MLP classifier implemented from scratch, comparing different optimization algorithms.

### The dataset

You are free to choose the dataset, as long as:
- it is a **classification** problem (binary or multi-class),
- it contains at least a few thousand datapoints,
- it is **not** Fashion-MNIST (which we used during the course).

Good places to look for a dataset are [Kaggle](https://www.kaggle.com/datasets), the [UCI Machine Learning Repository](https://archive.ics.uci.edu/) and [OpenML](https://www.openml.org/). Some examples of suitable datasets are: MNIST, KMNIST, EMNIST (images), *Adult Income*, *Bank Marketing*, *Credit Card Default*, *Wine Quality*, *Breast Cancer Wisconsin*, *Covertype* (tabular data). Datasets with images are simpler to pre-process, while tabular datasets often require more care (missing values, categorical features, unbalanced classes, ...): both are fine.

### What you have to do

Your notebook must cover **all** the steps of the Machine Learning pipeline we discussed during the course:

1. **The problem.** Briefly describe the dataset and the classification task: what are the inputs, what are the outputs, how many datapoints, features and classes are there?
2. **Data.** Download (or load) the data, explore it, and pre-process it in a way which is suitable for training a Neural Network with gradient-based methods. Justify each of your choices.
3. **Model.** Define an MLP classifier, implemented from scratch, and justify the choice of its architecture and of its loss function. Make sure that the computation of the gradient is correct.
4. **Training.** Train the model with **at least three** different optimization algorithms among the ones studied in the course (e.g. GD, SGD, SGD with Momentum, Adam). Discuss the choice of the hyperparameters of each of them.
5. **Comparison.** Compare the algorithms, using the plots and the metrics you think are the most informative, and discuss the results: which algorithm works best on your problem, and why? Is the comparison fair?
6. **Evaluation.** Evaluate your final model in a way which gives a reliable estimate of its ability to generalize to new data, and comment on its strengths and weaknesses.

There is no unique correct way to solve the project: different choices are fine, as long as they are **motivated**, and as long as the results are **correctly interpreted**.
